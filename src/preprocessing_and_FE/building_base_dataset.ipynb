```python
import pandas as pd
import numpy as np
```

```python
## Path Configuration


DEV_IN_PATH  = "../../data/raw/development.csv"
EVAL_IN_PATH = "../../data/raw/evaluation.csv"

DEV_OUT_PATH  = "../../data/processed/development_processed.csv"
EVAL_OUT_PATH = "../../data/processed/evaluation_processed.csv"

NUM_COLS = [
	"n_tokens",
	"title_len",
	"article_len",
	"title_ratio",
]

OUT_DIR.mkdir(parents=True, exist_ok=True)
```

```python
## Load data from CSV file
df_dev  = pd.read_csv(DEV_IN_PATH)
df_eval = pd.read_csv(EVAL_IN_PATH)


def load_and_basic_clean(path):
	df = pd.read_csv(path)

	df["article"] = df["article"].fillna("").astype(str)
	df["title"]   = df["title"].fillna("").astype(str)
	df["source"]  = df["source"].fillna("").astype(str)

	return df
```

```python
## Text FEATURE ENGINEERING

def build_text(df):
	df["text"] = (df["title"] + " " + df["article"]).str.lower()
	return df
```

```python
## Numerical FEATURE ENGINEERING
def add_numeric_features(df):
	df["n_tokens"]    = df["article"].str.split().str.len()
	df["title_len"]   = df["title"].str.len()
	df["article_len"] = df["article"].str.len()
	df["title_ratio"] = df["title_len"] / (df["article_len"] + 1)

	df[NUM_COLS] = (
		df[NUM_COLS]
		.replace([np.inf, -np.inf], 0)
		.fillna(0)
	)

	return df
```

```python
# Columns Selection
def preprocess(df):
	df = load_and_basic_clean(df)
	df = build_text(df)
	df = add_numeric_features(df)
	return df
```

```python
# Saving File  
# Processing DEV set
df_dev = load_and_basic_clean(DEV_IN_PATH)
df_dev = build_text(df_dev)
df_dev = add_numeric_features(df_dev)

print("DEV columns:")
print(df_dev.columns)
print("DEV shape:", df_dev.shape)

df_dev.to_csv(DEV_OUT_PATH, index=False)
print("Saved:", DEV_OUT_PATH)
# Processing EVAL set
df_eval = load_and_basic_clean(EVAL_IN_PATH)
df_eval = build_text(df_eval)
df_eval = add_numeric_features(df_eval)

print("EVAL columns:")
print(df_eval.columns)
print("EVAL shape:", df_eval.shape)

df_eval.to_csv(EVAL_OUT_PATH, index=False)
print("Saved:", EVAL_OUT_PATH)
```
